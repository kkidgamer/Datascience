```python
#visualisatiom
#matplotlib-library for plotting
#seaborn-library for high-level interf ace for drawing attractive statistical graphics
import seaborn as sns
import matplotlib as plt
plt.style.available
```

```python
plt.style.use("seaborn-v0_8")
```

```python
#pallets can be reds, oranges , magma ,flare, crest
#variable relationship are categorised by the number of variables analysed at a time

# univariate analysis-single variable-main goal is describe data,find patterns and summarised distribition-common plots-histogram,bi plot,piecharts

# bivariate-two variables simultaneously to find relationships,correlation association-common plots- scatter plot-two continous,bar plot-one categorical one continous,pair plot and line plots

# multivariate-three or more variables-to uncover common interactions-common plots grouped bar charts,swam/scatter plots,heatmap,3D scatter plots 
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(x=df["Gender"],palette="Greens")
plt.title("Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Members")
```

```python

sns.barplot(data=df,x="Gender",y="LoanAmount",palette="magma")
plt.title('Loan Amount by Gender')
plt.xlabel("Gender")
plt.ylabel("Loan Amount")
```

```python
sns.barplot(data=df,x="Gender",y="LoanAmount",hue="Education", palette="magma")
plt.title("Loan Amount by Gender and eduaction")
plt.xlabel("Gender")
plt.ylabel("Loan Amount")
```

```python
#swarmplot-visualize distribution of data points in a data set
sns.swarmplot(data=df,x="Gender",y="ApplicantIncome",hue="Self_Employed")
```
