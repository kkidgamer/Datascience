```python
import matplotlib.pyplot as plt
import seaborn as sns
students=["tamara","Hallima","kigen","john"]
marks=[70,20,45,90]
sns.barplot(x=students,y=marks,palette="plasma")
plt.savefig("students.png")
#viridis,plasma,magma,cool,coolwarm,Reds,Blues,Set1,Set2,Pastel1,bone,autumn,Greens,summer,winter,spring
```

```python

students=["Jayden","Tamara","Baraka","Kigen","John"]
#("students")
marks=[80,50,80,60,20]

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
plt.bar(students,marks,color="black")
plt.title("Students Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
#plt.show()
plt.grid()
plt.savefig("Graphs/studentsmarks.png")
```

```python
presidents_name=["Winny","Hassan","Mary","Trevor"]
Number_of_votes=[4000,9000,7000,8000]


plt.figure(figsize=(10,5))
plt.bar(presidents_name,Number_of_votes,color= "blue")
plt.title("Presidents")
plt.xlabel("presidents")
plt.ylabel("Number_of_votes")
plt.grid()
plt.savefig("Graphs/voters.png")
```

```python
temperature=[20,10.5,25,20,15.5]
time=[1,2,3,4,5]

plt.figure(figsize=(10,6))
plt.plot(time,temperature,color="Red",marker="o",linestyle="dashed")
plt.title('Temperature vs Time')
plt.xlabel("Time")
plt.ylabel("temperature")
plt.grid()
plt.savefig("Graphs/TempraturevsTime.png")
```

```python
#pie chart
#data
departments=["Human Resources","Marketing","Operations","Training"]
percentages=[50,20,20,10]
colors=["blue","yellow","orange","lightgreen"]
explode=(0.1,0.2,0.3,0.4) #explode the first slice

plt.figure(figsize=(20,5))
plt.pie(percentages,labels=departments,autopct="%1.1f%%",colors=colors,shadow=True,explode=explode)
plt.savefig("Graphs/DepartmentsDistribution.png")
```
