```python
#data visualisation-this the process of representing data in a graphical format
#it helps us understand the data better and to identify,patterns,trenda and outlier

```

```python
#histogram
import matplotlib.pyplot  as plt
import seaborn as sns

sns.histplot(data=df,x="Age",hue="Survived",palette="cool")
plt.savefig("Graphs/Agecount.jpeg")
```

```python
plt.hist(df["Age"],bins=10,edgecolor="black")
plt.title("Titanic tragedy")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.savefig("Graphs/Agecount2.jpeg")
```

```python
plt.scatter(df["Age"],df["Fare"])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare distribution")
plt.savefig("Graphs/AgevsFare.png")
```

```python
sns.scatterplot(x="Age",y="Fare",hue="Survived",data=df,palette="cool")
plt.savefig("Graphs/agevsfare.png")
```

```python
#piechart
gender=df["Sex"].value_counts()
plt.pie(gender.values,labels=gender,autopct="%1.1f%%",colors=["blue","brown"],explode=[0.1,0])
plt.savefig("Graphs/plot.png")
```
