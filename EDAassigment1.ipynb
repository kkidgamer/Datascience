```python
import matplotlib.pyplot as plt

# 1. Group by both County and Meter_Type, then unstack to create a table
meter_by_county = df.groupby(["County", "Meter_Type"]).size().unstack(fill_value=0)

# 2. Plot as a grouped bar chart
meter_by_county.plot(kind="bar", figsize=(10, 5))

# 3. Clean up the labels
plt.title("Meter Types by County", fontsize=14, fontweight='bold')
plt.xlabel("County")
plt.ylabel("Count")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```
